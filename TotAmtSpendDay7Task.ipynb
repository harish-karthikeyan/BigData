```python
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql.functions import round, sum
```

```python
spark = (
    SparkSession.builder.appName("TotAmtSpendDay7Task")
    .master("local[*]")
    .getOrCreate()
)
```

```python
pdf = pd.read_excel("/content/customers order.xlsx")
df = spark.createDataFrame(pdf)
df.printSchema()
```

```python
result_df = (
    df.groupBy("Order ID")
    .agg(round(sum("Total Sales"), 2).alias("Total Spent"))
    .orderBy("Order ID")
)
```
