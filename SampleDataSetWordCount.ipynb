```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder \
  .appName("SampleDataSetWordCount") \
  .master("local[*]") \
  .getOrCreate()
```

```python
sc = spark.sparkContext
rdd = sc.textFile("youtubecommentsdataset.csv")
```

```python
rdd.collect()
comments = rdd.map(lambda line: line.split(",",1)[0])
comments.collect()
```

```python
words = comments.flatMap(lambda line: line.lower().split())
word_pairs = words.map(lambda word: (word,1))
print(word_pairs.take(10))
```

```python
word_counts = word_pairs.reduceByKey(lambda a , b : a+  b )
word_counts.collect()
```

```python
sorted_word_counts = word_counts.sortBy(lambda x: x[1], ascending = False)
sorted_word_counts.collect()
```

```python
comments = rdd.map(lambda line: line.rsplit(",",)[0])
```

```python
import nltk
nltk.download('stopwords')

from nltk.corpus import stopwords
stopword_Set = set(stopwords.words('english'))
```

```python
import re
words = comments.flatMap(lambda line: re.findall(r'\w+', line.lower()))
filtered_words = words.filter(lambda word: word not in stopword_Set)
filtered_words.collect()
```

```python
cleaned_comments = comments.map(lambda line: " ".join(
    [word for word in re.findall(r'\w+', line.lower()) if word not in stopword_Set]
))
paragraph = " ".join(cleaned_comments.collect())
print(paragraph)
```

```python
word_counts = words.map(lambda word: (word,1))\
                    .reduceByKey(lambda a,b: a+b)
for word, count in word_counts.collect():
  print(f"{word} : {count}")
```
