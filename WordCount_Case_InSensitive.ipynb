```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder \
  .appName("CaseInSensitive") \
  .master("local[*]") \
  .getOrCreate()
```

```python
sc = spark.sparkContext
rdd = sc.parallelize(["Big Data", "big spark", "Data Science"])
words = rdd.flatMap(lambda line: line.lower().split()) #convert to lowercase
pairs = words.map(lambda word: (word, 1))
counts = pairs.reduceByKey(lambda a,b: a+b)
print(counts.collect())
```
