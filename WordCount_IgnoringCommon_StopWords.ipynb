```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder \
  .appName("StopWords") \
  .master("local[*]") \
  .getOrCreate()
```

```python
sc = spark.sparkContext
stopwords = {'the', 'is','a','of','and','to'}
rdd = sc.parallelize(["Big data is the future" , "The future of the data and AI"])
words = rdd.flatMap(lambda line: line.lower().split())
filtered = words.filter(lambda word: word not in stopwords)
pairs = filtered.map(lambda word: (word,1))
counts = pairs.reduceByKey(lambda a,b: a+b)
print(counts.collect())
```
