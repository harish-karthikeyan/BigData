```python
import pandas as pd
from pyspark.sql import SparkSession
```

```python
spark = (
    SparkSession.builder.appName("TotAmtSpdDay3Task")
    .master("local[*]")
    .getOrCreate()
)
```

```python
pdf = pd.read_excel("/content/customers order.xlsx")
```

```python
df = spark.createDataFrame(pdf)
```

```python
total_spent = (
    df.groupBy("Order ID")
    .sum("Total Sales")
    .withColumnRenamed("sum(Total Sales)", "Total_Spent")
)
```

```python
total_spent.orderBy("Total_Spent", ascending=False).show()
```
