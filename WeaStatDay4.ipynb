```python
from pyspark.sql import SparkSession
```

```python
spark = (
    SparkSession.builder.appName("WeatherStationMaxTemp")
    .master("local[*]")
    .getOrCreate()
)
```

```python
sc = spark.sparkContext
rdd = sc.textFile("/content/WeatherStation.csv")
```

```python
header = rdd.first()
rdd_no_header = rdd.filter(lambda row: row != header and len(row.strip()) > 0)
```

```python
def parse_temp(line):
    fields = line.split(",")
    if len(fields) >= 4:
        try:
            return float(fields[3].strip())
        except ValueError:
            return None
    return None
```

```python
temps_rdd = rdd_no_header.map(parse_temp).filter(lambda x: x is not None)
max_temp = temps_rdd.reduce(lambda a, b: max(a, b))
```

```python
print(f"Maximum Temperature Observed: {max_temp}°C")
```
