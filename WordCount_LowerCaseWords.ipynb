```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder \
  .appName("LowerCaseWords") \
  .master("local[*]") \
  .getOrCreate()
```

```python
sc= spark.sparkContext
rdd =sc.parallelize(["Big Data", "big spark", "Data Science", "data", "apple", "zebra"])
words = rdd.flatMap(lambda line: line.split())
lowercase_words = words.filter(lambda word: word[0].islower())
word_counts = lowercase_words.map(lambda word:(word,1)) \
                            .reduceByKey(lambda a,b : a + b)
print(word_counts.collect())
```
