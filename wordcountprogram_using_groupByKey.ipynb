```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder \
  .appName("groupByKey") \
  .master("local[*]") \
  .getOrCreate()
```

```python
sc = spark.sparkContext
rdd = sc.parallelize(["big data", "big spark", "data science"])
word_pairs = rdd.flatMap(lambda line: line.split()).map(lambda word: (word, 1))
grouped = word_pairs.groupByKey()
word_counts = grouped.mapValues(lambda values: sum(values))
print(word_counts.collect())
```
