```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, max as _max
```

```python
spark = (
    SparkSession.builder.appName("weatStaMaxTempDay6Task")
    .master("local[*]")
    .getOrCreate()
)
```

```python
df = (
    spark.read.option("header", "true")
    .option("inferSchema", "true")
    .csv("/content/WeatherStation.csv")
)
```

```python
df_cleaned = df.withColumn(
    "Temperature (C)", col("Temperature (C)").cast("double")
)
```

```python
max_temp_df = df_cleaned.groupBy("Summary").agg(
    _max("Temperature (C)").alias("Max_Temperature")
)
```

```python
max_temp_df.show()
```
