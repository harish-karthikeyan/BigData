```python
import re
from pyspark.sql import SparkSession
```

```python
spark = (
    SparkSession.builder.appName("WordCountDay2Task")
    .master("local[*]")
    .getOrCreate()
)
```

```python
sc = spark.sparkContext
rdd = sc.textFile("/Books.csv")
```

```python
def clean_and_split(line):
    return re.split(r"\W+", line)
```

```python
word_counts_sorted = (
    rdd.flatMap(clean_and_split)
    .map(lambda word: word.lower())
    .filter(
        lambda word: len(word) > 0
    )
    .map(lambda word: (word, 1))
    .reduceByKey(lambda a, b: a + b)
    .map(
        lambda x: (x[1], x[0])
    )
    .sortByKey(ascending=False)
    .map(lambda x: (x[1], x[0]))
)
```

```python
results = word_counts_sorted.take(20)
for word, count in results:
    print(f"{word}: {count}")
```
